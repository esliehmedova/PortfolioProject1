```python
import pandas as pd
from sqlalchemy import create_engine

# Replace with your own MySQL username and password
engine = create_engine("mysql+pymysql://user:password@localhost:3306/portfolioprojects")

deaths = pd.read_excel("CovidDeaths.xlsx")
vacc = pd.read_excel("CovidVaccinations.xlsx")

deaths.to_sql("coviddeaths", engine, if_exists="fail", index=False, chunksize=10000)
vacc.to_sql("covidvaccinations", engine, if_exists="fail", index=False, chunksize=10000)

print("Done!", len(deaths), len(vacc)) 
```
